Imports

```python
import pandas as pd
from pymongo import MongoClient
```

Mongodb Connect

```python
client = MongoClient("mongodb://localhost:27017/")
```

DB Creation

```python
db = client["ecommerce_recommendation"]
```

Create Collection

```python
reviews_collection = db["reviews"]
```

Loading Dataset

```python
df = pd.read_json(
    "../data/cleaned_dataset.json",
    lines=True
)
```

Convert dataframe → MongoDB documents

```python
records = df.to_dict(orient="records")
```

Adding embedded document

```python
for record in records:
    record["analytics"] = {
        "sentiment": record.pop("sentiment"),
        "review_length": record.pop("review_length"),
        "helpful_ratio": record.pop("helpful_ratio")
    }
```

Insertion

```python
for record in records:
    record.pop("_id", None)
reviews_collection.insert_many(records)
```
