Mongodb Connection

```python
from pymongo import MongoClient
```

```python
client = MongoClient("mongodb://localhost:27017/")
```

```python
db = client["ecommerce_recommendation"]
```

```python
reviews_collection = db["reviews"]
users_collection = db["users"]
products_collection = db["products"]
interactions_collection = db["interactions"]
```

Finding highly-rated purchases

```python
high_rated = reviews_collection.find(
    {
        "overall": {"$gte": 4},
        "user_action": "purchase"
    }
)
```

```python
for review in high_rated.limit(5):
    print(review)
```

```python
projected_reviews = reviews_collection.find(
    {},
    {
        "_id": 0,
        "reviewerID": 1,
        "asin": 1,
        "overall": 1,
        "user_action": 1
    }
)
```

```python
for review in projected_reviews.limit(5):
    print(review)
```

Getting most recent interactions

```python
recent_interactions = interactions_collection.find().sort(
    "timestamp", -1
)
```

```python
for interaction in recent_interactions.limit(5):
    print(interaction)
```

```python
page_2_reviews = reviews_collection.find().skip(10).limit(5)
```

```python
for review in page_2_reviews:
    print(review)
```

Finding low-rated products

```python
bad_products = reviews_collection.find(
    {
        "overall": {"$lt": 3}
    }
)
```

```python
for review in bad_products.limit(5):
    print(review)
```

Finding neutral reviews

```python
neutral_reviews = reviews_collection.find(
    {
        "overall": {"$eq": 3}
    }
)
```

```python
for review in neutral_reviews.limit(5):
    print(review)
```

Non-purchase interactions

```python
non_buyers = interactions_collection.find(
    {
        "action": {"$ne": "purchase"}
    }
)
```

```python
for users in non_buyers.limit(5):
    print(users)
```

Finding Valuable Actions

```python
valuable_actions = interactions_collection.find(
    {
        "action": {
            "$in": ["purchase", "add_to_cart"]
        }
    }
)
```

```python
for interaction in valuable_actions.limit(5):
    print(interaction)
```

Ignoring Weak Interactions

```python
strong_interactions = interactions_collection.find(
    {
        "action": {
            "$nin": ["ignore", "view"]
        }
    }
)
```

```python
for interaction in strong_interactions.limit(5):
    print(interaction)
```

Finding highly positive and helpful reviews

```python
trusted_reviews = reviews_collection.find(
    {
        "$and": [
            {"overall": {"$gte": 4}},
            {"analytics.helpful_ratio": {"$gte": 0.5}}
        ]
    }
)
```

```python
for review in trusted_reviews.limit(5):
    print(review)
```

Finding Negative / Neutral Sentiment

```python
negative_feedback = reviews_collection.find(
    {
        "$or": [
            {"analytics.sentiment": "negative"},
            {"analytics.sentiment": "neutral"}
        ]
    }
)
```

```python
for review in negative_feedback.limit(5):
    print(review)
```

Finding Not Highly-Rated Products

```python
not_high_rated = reviews_collection.find(
    {
        "overall": {
            "$not": {
                "$gte": 4
            }
        }
    }
)
```

```python
for review in not_high_rated.limit(5):
    print(review)
```

Marking extremely helpful reviews as “featured”

```python
try:
    res = reviews_collection.update_many(
        {
            "analytics.helpful_ratio": {"$gte": 0.8}
        },

        {
            "$set": {
                "featured_review": True
            }
        },
        upsert=True
    )
    if hasattr(res, 'upserted_id') and res.upserted_id:
        print(f"Inserted new record with ID: {res.upserted_id}")
    else:
        print(f"{res.modified_count} records updated.")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
reviews_collection.find_one(
    {
        "featured_review": True
    }
)
```

Flagging suspiciously short negative reviews

```python
try:
    res = reviews_collection.update_many(
        {
            "$and": [
                {"analytics.sentiment": "negative"},
                {"analytics.review_length": {"$lt": 15}}
            ]
        },

        {
            "$set": {
                "possible_spam": True
            }
        },
        upsert=True
    )
    if hasattr(res, 'upserted_id') and res.upserted_id:
        print(f"Inserted new record with ID: {res.upserted_id}")
    else:
        print(f"{res.modified_count} records updated.")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
reviews_collection.find_one(
    {
        "possible_spam": True
    }
)
```

Delete Operations

```python
from datetime import datetime, UTC
```

Soft deleting suspicious spam reviews

```python
try:
    res = reviews_collection.update_many(
        {
            "possible_spam": True
        },

        {
            "$set": {
                "deleted_at": datetime.now(UTC)
            }
        },
        upsert=True
    )
    if hasattr(res, 'upserted_id') and res.upserted_id:
        print(f"Inserted new record with ID: {res.upserted_id}")
    else:
        print(f"{res.modified_count} records updated.")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
reviews_collection.find_one(
    {
        "deleted_at": {"$exists": True}
    }
)
```

Hard deleting empty/useless reviews

```python
try:
    res = reviews_collection.delete_many(
        {
            "analytics.review_length": {"$lte": 2}
        }
    )
    if res.deleted_count == 0:
        raise ValueError("Delete operation found no records matching the query.")
    else:
        print(f"{res.deleted_count} records deleted.")
except ValueError as ve:
    print(f"Warning: {ve}")
except Exception as e:
    print(f"An error occurred: {e}")
```

Removing one extremely low-value interaction.

```python
try:
    res = interactions_collection.delete_one(
        {
            "action": "ignore",
            "rating": 1
        }
    )
    if res.deleted_count == 0:
        raise ValueError("Delete operation found no record matching the query.")
    else:
        print(f"{res.deleted_count} records deleted.")
except ValueError as ve:
    print(f"Warning: {ve}")
except Exception as e:
    print(f"An error occurred: {e}")
```
