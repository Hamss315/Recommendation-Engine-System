Connecting to Mongodb

```python
from pymongo import MongoClient
```

```python
client = MongoClient("mongodb://localhost:27017/")
```

```python
db = client["ecommerce_recommendation"]
```

Getting Collections

```python
reviews_collection = db["reviews"]
users_collection = db["users"]
products_collection = db["products"]
interactions_collection = db["interactions"]
```

Single-field indexes

```python
interactions_collection.create_index("user_id")
```

```python
interactions_collection.create_index("product_id")
```

```python
reviews_collection.create_index("overall")
```

```python
print(interactions_collection.index_information())
```

Compound indexes

```python
interactions_collection.create_index([
    ("user_id", 1),
    ("product_id", 1)
])
```

```python
reviews_collection.create_index([
    ("asin", 1),
    ("overall", -1)
])
```

```python
print(interactions_collection.index_information())
```

```python
print(reviews_collection.index_information())
```

Multi-Key indexes

```python
users_collection.create_index("interaction_ids")
```

```python
products_collection.create_index("interaction_ids")
```

```python
print(users_collection.index_information())
print(products_collection.index_information())
```
