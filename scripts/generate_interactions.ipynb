Imports

```python
from pymongo import MongoClient
import random
```

Mongodb Connect

```python
client = MongoClient("localhost", 27017)
db = client["ecommerce_recommendation"]
```

```python
reviews_collection = db["reviews"]
```

```python
reviews = list(reviews_collection.find())
```

Creating Interactions

```python
interactions = []
for review in reviews:
    interaction = {

        "user_id": review["reviewerID"],

        "product_id": review["asin"],

        "action": review["user_action"],

        "rating": review["overall"],

        "timestamp": review["unixReviewTime"]
    }
    interactions.append(interaction)
```

```python
interactions_collection = db["interactions"]
interactions_collection.drop()
interactions_collection.insert_many(interactions)
```

```python
print(interactions_collection.count_documents({}))
```

```python
interactions_collection.find_one()
```

Creating Users

```python
users = []
unique_users = reviews_collection.distinct("reviewerID")
for user in unique_users:
    user_doc = {
        "user_id": user
    }
    users.append(user_doc)
```

```python
users_collection = db["users"]
```

```python
users_collection.drop()
```

```python
users_collection.insert_many(users)
```

```python
print(users_collection.count_documents({}))
```

```python
users_collection.find_one()
```

Creating Products

```python
products = []
unique_products = reviews_collection.distinct("asin")
for product in unique_products:
    product_doc = {
        "product_id": product
    }
    products.append(product_doc)
```

```python
products_collection = db["products"]
```

```python
products_collection.drop()
```

```python
products_collection.insert_many(products)
```

```python
print(products_collection.count_documents({}))
```

```python
products_collection.find_one()
```

Interaction References inside Users

```python
all_interactions = list(interactions_collection.find())
```

```python
user_interactions = {}
```

Building Mapping Users & Interactions

```python
for interaction in all_interactions:
    user_id = interaction["user_id"]
    if user_id not in user_interactions:
        user_interactions[user_id] = []
    user_interactions[user_id].append(
        interaction["_id"]
    )
```

```python
users_collection.drop()
```

Updating Users Collection

```python
users = []
for user_id, interaction_ids in user_interactions.items():
    user_doc = {
        "user_id": user_id,
        "interaction_ids": interaction_ids
    }
    users.append(user_doc)
```

```python
users_collection.drop()
```

```python
users_collection.insert_many(users)
```

```python
users_collection.find_one()
```

Building Mapping Products & Interactions

```python
product_interactions = {}
```

```python
for interaction in all_interactions:
    product_id = interaction["product_id"]
    if product_id not in product_interactions:
        product_interactions[product_id] = []
    product_interactions[product_id].append(
        interaction["_id"]
    )
```

```python
products = []
```

```python
for product_id, interaction_ids in product_interactions.items():
    product_doc = {
        "product_id": product_id,
        "interaction_ids": interaction_ids
    }
    products.append(product_doc)
```

```python
products_collection.drop()
```

Updating Products Collection

```python
products_collection.insert_many(products)
```

```python
products_collection.find_one()
```
